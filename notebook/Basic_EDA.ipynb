### Checking Outliers 

```python
numerical_columns = df.select_dtypes(('int64','float64'))
for col in numerical_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    
    print(f"{col}: {len(outliers)} outliers")
```

### Observation:
- The Dataset contain outliers in 4 columns and the columns name is:
- age: 60 outliers
- income: 2190 outliers
- browsing_time: 1313 outliers
- pages_viewed: 1262 outliers
- previous_purchases: 445 outliers
