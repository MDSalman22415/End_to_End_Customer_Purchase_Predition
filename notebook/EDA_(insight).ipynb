## UNIVARIATE ANALYSIS IN CATEGORY DATA

```python
category_columns = df.select_dtypes(('object'))
```

```python
for col in category_columns:

    sns.countplot(data=df, x=col, palette="Set1")

    plt.xlabel(col)
    plt.ylabel("Counts")
    plt.title(f"Distribution of {col}")
    plt.xticks(rotation=45)

    plt.show()
```

### Observation:

* The number of **female customers is higher than male customers**.
* The **Fashion category has the highest number of customers**.
* A **smaller number of customers used a discount**.

```python
for col in category_columns:

    counts = df[col].value_counts

    plt.pie(
        counts,
        labels=counts.index,
        autopct='%1.1f%%'
    )

    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel("")
    plt.show()
```

### Observation:
- In Gender Columns 49.2% customer is  Female and 47.7% customer  Male in conclusion higest number of customer is Female 
- In Product Category higest number of customer in Fashion category Around 21.9%
- Small number of customer to used a discount around 41.7%

### Which Product_category has the highest number of customers by gender

```python
customer = pd.crosstab(df['gender'], df['product_category'])
```

### Conclusion

1. **Fashion is the most popular product category**, and the highest number of customers purchase products from this category.

2. **Female customers have a higher average income and higher average page views than male customers.** This indicates that female customers are more active on the e-commerce website.

3. **Customers who receive discounts have a higher purchase rate.** This shows that discounts can encourage customers to purchase products.

4. **Customers with previous purchases are slightly more likely to purchase again**, but previous purchase history does not have a very strong relationship with the current purchase.

5. **Average income is almost similar across customers**, although the median income may show some differences.

6. Overall, **discounts, product category, and customer activity (page views) appear to be useful factors related to purchasing behavior**, while previous purchases have only a weak relationship with current purchases.
