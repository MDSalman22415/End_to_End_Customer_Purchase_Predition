```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('../data/customer_purchase_merchant_dataset_50000.csv')
```

### To Correct Inconsistent values

```python
df['gender'].value_counts()
```

```python
map_gender = {
    'Male' : 'Male',
    'Female':'Female',
    'FEMALE':'Female',
    'Other': 'Other',
    'male' :'Male',
    'other':'Other',
    'Unknown':'Unknown',
    'X':'Male'
}
df['gender'] = df['gender'].map(map_gender)
```

```python
df['gender'].value_counts()
```

```python
product_category_map = {
    'Fashion' :  'Fashion',
    'Grocery' :  'Grocery',
    'Home'    :   'Home',
    'Beauty'  :   'Beauty',
    'Electronics' :'Electronics',
    'Sports' : 'Sports',
    'electronics' :'Electronics',
    'grocery' :  'Grocery',
    'HOME' :     'Home',
    'FASHION' : 'Fashion',
    'SPORTS' :'Sports',
    'beauty' :  'Beauty'
} 
df['product_category']  = df['product_category'].map(product_category_map)
```

```python
df['product_category'].value_counts()
```

```python
df.drop_duplicates(inplace=True)
```
